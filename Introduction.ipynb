* Statistics is the process of collecting, organizing, analyzing, interpreting, and communicating data.

## Data and Variables

* A **variable** is a characteristic that can have different values For example: Variable Age, Salary, Gender, can have different values. Data is information collected for a study or analysis.

## Types of Data

### Qualitative & Quantitative Data

1. * Qualitative Data: Describes qualities, groups, or categories. for example: Eye color, favorite food, or pass/fail.
2. * Quantitative Data: Deals with numbers, amounts, and measurements for example: money earned, Height, temperature.

Under Qualitative data we have:

* Nominal Data - represents categories with no specific order or ranking (such as eye color or Gender).
* Ordinal Data - containing categories with a meaningful order or rank(such as low, medium, and high income, poor, good, great)

Under quantitative data we have:

* Discrete Data - which contain whole numbers/numbers that are countable. eg Number of students in a class can be 25, and not 25.4, Goals scored in a soccer match can be 3 and not 3.5
* Continuous Data - values that represent measurements and can take any numerical value within a range, including fractions and decimals. Examples include a person's height, weight (175.5 cm, 65.4 kg) or a temperature reading (22.4°C). 


## Population and Sample

* Suppose a company has 10,000 employees. The entire 10,000 employees are the **Population**
* If we select 200 employees from the 10,000 population then those 200 employees are the **Sample**

* We use samples because studying an entire population can be impossible, time-consuming and expensive. So a carefully selected sample allows us to learn about the entire population.

## Parameter & Statistic

*  A parameter is a numerical value that describes an entire population for example The average salary of all 10,000 employees in a company. Whereas, a statistic is a numerical value that describes a small sample taken from that population i.e  the average salary of 200 selected employees is a statistic.

## Types of Statistics

### Descriptive, Inferential Statistics

Descriptive statistics describes, organizes and summarizes the data we have. This is achieved through these methods: 
* Measures of Central Tendency: Find the "center" or a typical value of your data (e.g., Mean, Median, Mode most common number).
* Measures of Dispersion: Show how spread out the data points are (e.g., Range, Variance, Standard Deviation, and Interquartile Range (IQR))
* Measures of Distribution Shape: Show how the data is shaped or skewed (e.g., Skewness, Kurtosis, and Percentiles).
* Visual/Graphical Tools to summarize data visually using Histograms, Bar Charts, Pie Charts, Scatter Plots, and Box Plots.

Inferential statistics uses data from a sample to make predictions, generalize, or draw conclusions about a larger population. This can be achieved through these methods:
* Hypothesis Testing: Testing a specific assumption about a population (e.g., t-tests, ANOVA, and Chi-Square tests).
* Confidence intervals: An educated guess about a large group based on a small sample. Instead of guessing a single, exact number, it gives us a safe range where the real answer is likely to be.
* Regression & Correlation Analysis: Exams connections between different things to see how they affect each other. Correlation checks if two things change together (like how height and weight relate)while regression measure  relationship mathematically (i.e using Linear Regression) to predict a future outcome based on current data.
* Regression & Correlation Analysis: Examines relationships between different variables to understand how they are associated with one another. **Correlation** measures whether two variables change together, such as the relationship between height and weight. **Regression** models the relationship between variables mathematically and can be used to predict future outcome based on current data.
